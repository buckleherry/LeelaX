```python
import chess, torch
from leelax.net.model import LeelaXNet
from leelax.mcts.puct import PUCT
from leelax.env.policy_index import index_to_move

net = LeelaXNet().to("cpu")
def net_fn(x):
    with torch.no_grad():
        return net(x)

board = chess.Board()
mcts = PUCT(net_fn, n_simulations=32)
policy, action_idx = mcts.run(board)

print("policy sum:", policy.sum())
mv = index_to_move(board, action_idx)
san = board.san(mv)
uci = mv.uci()
print(f"chosen move: {uci} | {san}")
```
