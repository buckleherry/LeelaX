```python
from leelax.net.model import LeelaXNet
from leelax.selfplay.worker import SelfPlayWorker
from leelax.selfplay.replay_buffer import ReplayBuffer
import torch

net = LeelaXNet()
def net_fn(x):
    with torch.no_grad():
        return net(x)

# n_simulations/max_moves kept small so the demo runs quickly
worker = SelfPlayWorker(net_fn, n_simulations=16, max_moves=60)
buffer = ReplayBuffer(capacity=50_000)

# play 5 games and store them
for _ in range(5):
    samples, moves = worker.play_game(verbose=False)
    buffer.add_many(samples)

print("buffer size:", len(buffer))

# draw a training batch
states, policies, values = buffer.sample(batch_size=32)
print(states.shape, policies.shape, values.shape)
```
