```python
from leelax.env.encode import state_to_tensor
from leelax.env.policy_index import legal_policy_mask
from leelax.net.model import LeelaXNet
from leelax.net.inference import masked_softmax, select_action
import chess, torch

board = chess.Board()
state = state_to_tensor(board).unsqueeze(0)  # [1,24,8,8]
mask = torch.from_numpy(legal_policy_mask(board)).unsqueeze(0)  # [1,4672]

net = LeelaXNet()
policy_logits, value = net(state)
probs = masked_softmax(policy_logits, mask)
idx = select_action(policy_logits, mask, greedy=True)
```
