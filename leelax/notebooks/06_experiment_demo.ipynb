```python
import torch
from leelax.net.model import LeelaXNet
from leelax.selfplay.worker import SelfPlayWorker
from leelax.selfplay.replay_buffer import ReplayBuffer

model = LeelaXNet()
def net_fn(x):
    with torch.no_grad():
        return model(x)

worker = SelfPlayWorker(net_fn, n_simulations=32)

for i in range(3):
    samples, moves = worker.play_game(verbose=True)
    print(f"Game {i+1}: {len(moves)} moves, result {samples[0][2]}")
```
